```python
# Funzione per leggere e salvare i dati in una lista

import matplotlib.pyplot as plt
import numpy as np
#import pandas as pd

def from_file_to_list(file_path):

    # Inizializza una lista per salvare i dati
    dati = []

    # Prova ad aprire il file e leggere i dati
    try:
        with open(file_path, 'r') as file:

            # Leggi ogni riga del file
            for line in file:

                # Dividi la riga in base al separatore "\t"
                elements = line.strip().split('\t')  #line strip elimina spazi bianchi

                # Aggiungi i dati alla lista
                dati.append(elements)

    except FileNotFoundError:
        print("Il file non è stato trovato.")

    return dati

# La lista dati è stata creata
```

## **Leggere il file prodotto da CoolTerm e stampare i primi dati come check**


```python
import sys, os

# Inserisci il percorso della cartella che contiene i file creati da CoolTerm
LOCALPATH = './'

# Inserisci il nome del file di testo creato da CoolTerm
CoolTermFile = 'data_2024_04_30_B.txt'

# Inserisci il nome del file di testo con i dati in formato decimale per l'analisi
DataFile = 'data_for_analysis_B.txt'

# Chiamata alla funzione per leggere i dati
file_path = LOCALPATH + CoolTermFile

dati_list = from_file_to_list(file_path)

# Esempio di stampa dei primi 5 elementi per verificare se i dati sono stati letti correttamente
print("Primi 5 elementi dei dati salvati come lista:")
for row in dati_list[:5]:
    print(row)
```

## **Conversione da microsecondi in formato esadecimale a secondi in formato decimale**

```python
import re

# Converti ogni elemento di dati_list in una stringa
dati_list = [str(item) for item in dati_list]

# Inizializza una lista per i risultati
results = []
current_time = 0
counter = 0

# Analizza ogni riga dell'output
for line in dati_list:
    # Trova tutti i match di 't' seguiti da un numero esadecimale
    matches = re.findall(r't([0-9A-Fa-f]+)', line)

    # Se ci sono match, aggiungi il tempo e il numero di conteggi ai risultati
    if matches:
        for match in matches:
            time = int(match, 16)  # Converti il numero esadecimale in decimale
            bin_time = time * 10**(-6)  # Calcola il tempo in microsecondi all'interno del bin dove avviene il conteggio
            counter += 1  # Incrementa il contatore
            results.append((current_time + bin_time, counter))

# incrementa il tempo di un secondo 
    current_time += 1

# Scrivi i risultati su un file
with open(DataFile, "w") as file:
    for result in results:
        # Scrivi i tempi e i conteggi come due colonne separate
        file.write(f"{result[1]}\t{result[0]}\n")

# Il file è di due colonne:
# conteggi totali     tempo[s]
```
