```python
import matplotlib.pyplot as plt
import numpy as np

# Configurazione del grafico 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# 1. Disegniamo l'asse Z di rotazione (passante per l'origine)
z_line = np.linspace(-1.5, 2, 100)
ax.plot(np.zeros_like(z_line), np.zeros_like(z_line), z_line, 
        color='blue', linestyle='--', linewidth=2, label='Asse di rotazione Z')

# 2. Generiamo un corpo piatto bidimensionale sul piano XY (forma a uovo/irregolare)
theta = np.linspace(0, 2*np.pi, 200)
# Raggio variabile per dare una forma irregolare
r_body = 1.2 + 0.3 * np.sin(3 * theta)  
x_body = r_body * np.cos(theta)
y_body = r_body * np.sin(theta)
z_body = np.zeros_like(x_body) # Z = 0 perché giace sul piano XY

# Disegniamo la superficie del corpo piatto
ax.plot_trisurf(x_body, y_body, z_body, color='lightgray', alpha=0.6)
ax.plot(x_body, y_body, z_body, color='gray', linewidth=1.5)

# 3. Definiamo la posizione della particella P_i
px, py, pz = 0.8, 1.1, 0.0

# 4. Disegniamo le proiezioni (i cateti del triangolo)
# Proiezione lungo l'asse X
ax.plot([0, px], [0, 0], [0, 0], color='red', linewidth=2.5, label='Coordinata x')
ax.plot([px, px], [0, py], [0, 0], color='green', linestyle=':', linewidth=2)

# Proiezione lungo l'asse Y
ax.plot([0, 0], [0, py], [0, 0], color='green', linewidth=2.5, label='Coordinata y')
ax.plot([0, px], [py, py], [0, 0], color='red', linestyle=':', linewidth=2)

# 5. Disegniamo l'ipotenusa (rho_i): la vera distanza perpendicolare dall'asse Z
ax.plot([0, px], [0, py], [0, 0], color='blue', linewidth=3, 
        label=r'$\rho_i = \sqrt{x^2 + y^2}$ (Distanza da Z)')

# 6. Disegniamo la particella P_i (Punto Rosso)
ax.scatter([px], [py], [pz], color='red', s=150, zorder=5, label='Particella $P_i$')

# Testi e label grafiche
ax.text(px, py, 0.1, f'  P_i ({px}, {py})', color='black', fontsize=12, weight='bold')
ax.text(px/2, -0.15, 0, f'x = {px}', color='red', fontsize=11, weight='bold')
ax.text(-0.25, py/2, 0, f'y = {py}', color='green', fontsize=11, weight='bold')
ax.text(px/2 - 0.1, py/2 + 0.1, 0.1, r'$\rho_i$', color='blue', fontsize=14, weight='bold')

# Impostazioni degli assi e della vista
ax.set_xlabel('Asse X', color='red', fontsize=12)
ax.set_ylabel('Asse Y', color='green', fontsize=12)
ax.set_zlabel('Asse Z', color='blue', fontsize=12)
ax.set_title('Visualizzazione Geometrica di $\\rho_i^2 = x^2 + y^2$ (Corpo su piano XY)', fontsize=14)

# Limiti degli assi per una visualizzazione proporzionata
ax.set_xlim([-1.5, 1.5])
ax.set_ylim([-1.5, 1.5])
ax.set_zlim([-1, 1.5])

# Orientazione iniziale della telecamera 3D (puoi ruotarla col mouse!)
ax.view_init(elev=30, azim=-50)

ax.legend(loc='upper left')
plt.grid(True)
plt.show()
```
