```python
import numpy as np
import matplotlib.pyplot as plt

L = 0.4 # m
M = 5    # kg
m = 1.3   # kg

PiantaPiccolaPeso = 0.638
PiantaGrandePeso = 0.464
PiantaGrandeGrandePeso = 2.4

d = np.linspace(0.0001, L, 40000)
h = 0.20

g = 9.81

theta = np.arctan(h/d)

T = (M + m) * g * L / (4 * d * np.sin(theta))

print(f"Tensione minima: {np.min(T)} N")
print(f"Peso che il filo deve sostenere: {np.min(T) / g} kg")

T_lim = (M + m) * g * L / (4 * h)

plt.plot(d, T, label="Tensione singolo filo")
plt.axhline(T_lim, color='red', linestyle='--', label=fr"Tensione asintotica: ${T_lim}$")

plt.xlabel("Distanza dal perno (m)")
plt.ylabel("Tensione (N)")
plt.title("Tensione in funzione della distanza")
plt.grid()

plt.ylim(0, 500)
plt.legend()

plt.show()
```
