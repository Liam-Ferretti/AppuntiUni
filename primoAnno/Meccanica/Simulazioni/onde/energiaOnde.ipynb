$U(x, t) = \frac{1}{2}\sigma \omega_n^2 A_n^2\left[\sin^2(k_nx)\cos^2(\omega_nt) + \cos^2(k_nx)\sin^2(\omega_nt)\right]$

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
```

```python
# Parametri iniziali
A = 10
L = 10
v = 3
omega1 = np.pi * v / L
k1 = omega1 / v

x = np.linspace(0, L, 100)

# Creazione della figura e di 3 sottografici (3 righe, 1 colonna)
# sharex=True fa in modo che condividano lo stesso asse x
fig, axs = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

for n in range(1, 4):
    omega_n = n * omega1
    k_n = n * k1
    
    # Seleziona il sottografico corrente (n-1 perché gli indici partono da 0)
    ax = axs[n-1] 

    # --- Calcoli per t = 0 ---
    t1 = 0
    E_k1 = np.sin(k_n*x)**2 * np.cos(omega_n*t1)**2
    E_p1 = np.cos(k_n*x)**2 * np.sin(omega_n*t1)**2
    
    # Plot per t = 0 (linee continue)
    ax.plot(x, E_k1, '-', linewidth=2, label='E. cinetica (t=0)')
    ax.plot(x, E_p1, '-', linewidth=2, label='E. potenziale (t=0)')

    # --- Calcoli per t = pi / (2 * omega_n) ---
    t2 = np.pi / (2 * omega_n)
    E_k2 = np.sin(k_n*x)**2 * np.cos(omega_n*t2)**2
    E_p2 = np.cos(k_n*x)**2 * np.sin(omega_n*t2)**2

    # Plot per t > 0 (linee tratteggiate)
    ax.plot(x, E_k2, '--', linewidth=2, label=r'E. cinetica (t=$\pi/2\omega_n$)')
    ax.plot(x, E_p2, '--', linewidth=2, label=r'E. potenziale (t=$\pi/2\omega_n$)')

    # Dettagli del singolo sottografico
    ax.set_title(f'Modo n = {n}')
    ax.set_ylabel('Energia (J)')
    ax.legend(loc='upper right', fontsize='small')
    ax.grid(True)

# Imposta l'etichetta dell'asse x solo per l'ultimo grafico in basso
axs[-1].set_xlabel('Posizione lungo la corda (m)')

# Titolo generale per l'intera figura
fig.suptitle("Distribuzione dell'energia in una corda vibrante nei vari modi", fontsize=18)

# Regola gli spazi per evitare accavallamenti tra i grafici
plt.tight_layout()
plt.savefig("../Appunti/img/energiaOnde.png", dpi=300)
plt.show()
```
