```python
import matplotlib.pyplot as plt
import numpy as np

A = 10
L = 1
v = 3
omega1 = np.pi * v / L
k1 = omega1 / v

# Array per lo SPAZIO (da 0 a L)
x = np.linspace(0, L, 1000)

# Creiamo una figura con 3 grafici impilati verticalmente (3 righe, 1 colonna)
# sharex=True allinea perfettamente l'asse delle ascisse
fig, assi = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

colori = ['b', 'g', 'r']

# Cicliamo per disegnare le prime 3 armoniche
for n in [1, 2, 3]:
    k_n = n * k1
    omega_n = n * omega1  # Correzione: usiamo l'omega dell'armonica n-esima
    
    ax = assi[n-1]  # Selezioniamo il grafico corrente (indice da 0 a 2)
    
    # Riduciamo il numero di istanti temporali e usiamo alpha (trasparenza)
    # per creare un bell'effetto "corda vibrante" senza generare un blocco di colore solido
    tempi = np.linspace(0, 2, 30) 
    
    for t in tempi:
        xi = A * np.sin(k_n * x) * np.sin(omega_n * t) 
        ax.plot(x, xi, color=colori[n-1], alpha=0.3)
        
    # Disegniamo l'inviluppo massimo (i bordi dell'onda) con una linea nera tratteggiata
    ax.plot(x, A * np.sin(k_n * x), color='black', linestyle='--', linewidth=1.5)
    ax.plot(x, -A * np.sin(k_n * x), color='black', linestyle='--', linewidth=1.5)

    # Formattazione del singolo grafico
    ax.set_title(f"Armonica n={n}")
    ax.set_ylabel("xi")
    ax.grid(True)

# L'etichetta dell'asse X serve solo sull'ultimo grafico in basso
assi[2].set_xlabel("Posizione (x) lungo la corda")

# Aggiusta automaticamente gli spazi per evitare accavallamenti tra titoli e assi
plt.savefig("../Appunti/img/armonicheCorda.png", dpi=300)
plt.tight_layout()
plt.show()
```
