```python
import matplotlib.pyplot as plt
import numpy as np

A = 10
v = 1
w = 1
k = w / v

t = np.linspace(-20, 0, 1000)

plt.figure(figsize=(10, 6))

# Scegliamo una sfumatura (es. 'viridis' o 'plasma')
cmap = plt.get_cmap('coolwarm')
colors = [cmap(i) for i in np.linspace(0, 1, 31)]

for i, x_val in enumerate(range(0, 30, 1)):
    x = x_val / 10
    xi = A * np.cos(k * x - w * t) + A * np.cos(k * x + w * t)
    # Coloriamo la linea in base a x senza aggiungere label
    plt.plot(t, xi, color=colors[i])

# --- AGGIUNTA NODI E VENTRI ---

# Disegniamo il ventre (x = 0)
x_ventre = 0
xi_ventre = A * np.cos(k * x_ventre - w * t) + A * np.cos(k * x_ventre + w * t)
plt.plot(t, xi_ventre, color='black', linewidth=3, label=f'Ventre (x = {x_ventre})')

# Disegniamo il nodo (x = pi / 2)
x_nodo = np.pi / 2
xi_nodo = A * np.cos(k * x_nodo - w * t) + A * np.cos(k * x_nodo + w * t)
plt.plot(t, xi_nodo, color='black', linestyle='--', linewidth=3, label=f'Nodo (x ≈ {x_nodo:.2f})')

# ------------------------------

plt.title('Onda Stazionaria')
plt.xlabel('Tempo (t)')
plt.ylabel(r'Spostamento (\xi)')

# Creiamo una barra laterale per i valori di x
sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=0, vmax=2.9))
sm.set_array([])
cbar = plt.colorbar(sm, ax=plt.gca())
cbar.set_label('Posizione (x)')

# Aggiungiamo la legenda per rendere visibili il nodo e il ventre
plt.legend(loc='upper right')

plt.grid()

# Assicurati che le cartelle esistano se usi un percorso relativo
# plt.savefig('../Appunti/img/onde_stazionarie.png', dpi=300) 
plt.show()
```
