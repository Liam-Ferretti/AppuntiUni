```python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline

v = 0.7

# Frequenze angolari (tienile vicine per vedere bene il battimento)
omega1 = 1.05
omega2 = 1

# 2. Formula corretta per il vettore d'onda (k = omega / v)
k1 = omega1 / v
k2 = omega2 / v

# Delta e valori medi
dOmega = omega2 - omega1
omega0 = (omega2 + omega1) / 2

dk = k2 - k1
k0 = (k2 + k1) / 2

A = 4
# Asse del tempo
t = np.linspace(0-100, 100, 100000)

# Funzione d'onda risultante (Battimento)
f_tempo = 2 * A * np.cos(dOmega / 2 * t) * np.cos(omega0 * t)

# Inviluppo dell'onda (la linea tratteggiata che racchiude l'onda)
# Serve per vedere visivamente l'effetto del battimento
inviluppo_superiore_tempo = 2 * A * np.cos(dOmega / 2 * t)
inviluppo_inferiore_tempo = -2 * A * np.cos(dOmega / 2 * t)


# Asse del tempo
x = np.linspace(-100, 100, 100000)

# Funzione d'onda risultante (Battimento)
f_spazio = 2 * A * np.cos(dk / 2 * x) * np.cos(k0 * x)

# Inviluppo dell'onda (la linea tratteggiata che racchiude l'onda)
# Serve per vedere visivamente l'effetto del battimento
inviluppo_superiore_spazio = 2 * A * np.cos(dk / 2 * x)
inviluppo_inferiore_spazio = - inviluppo_superiore_spazio
```

```python
# Creazione del grafico
plt.figure(figsize=(10, 5))
plt.plot(t, f_tempo, label="Battimento", color="blue", alpha=0.7)

# Disegniamo anche l'inviluppo per chiarire il grafico (Opzionale, puoi cancellarlo)
plt.plot(
    t,
    inviluppo_superiore_tempo,
    label="Inviluppo",
    color="red",
    linestyle="--",
    alpha=0.5,
)
plt.plot(t, inviluppo_inferiore_tempo, color="red", linestyle="--", alpha=0.5)

plt.title(f"Battimento")
plt.xlabel("Tempo (t)")
plt.ylabel("Ampiezza")
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Creazione del grafico
plt.figure(figsize=(10, 5))
plt.plot(x, f_spazio, label="Battimento", color="blue", alpha=0.7)

# Disegniamo anche l'inviluppo per chiarire il grafico (Opzionale, puoi cancellarlo)
plt.plot(
    x,
    inviluppo_superiore_spazio,
    label="Inviluppo",
    color="red",
    linestyle="--",
    alpha=0.5,
)
plt.plot(x, inviluppo_inferiore_spazio, color="red", linestyle="--", alpha=0.5)

plt.title(f"Battimento")
plt.xlabel("Spazio (m)")
plt.ylabel("Ampiezza")
plt.grid(True)
plt.legend()

plt.show()
```
