```python
import matplotlib.pyplot as plt
import numpy as np

r = np.linspace(0.001, 0.05, 1000)

G = 10**-6
M = 1000
m = 2

# Energia potenziale gravitazionale
V_r = - G * M * m / r

# Momento angolare
P = 0

# Energia potenziale centrifuga (in questo caso 0)
V_c = P**2 / (2 * m * r**2)

# Energia potenziale efficace
V_eff = V_c + V_r

# 1. Impostiamo E_M minore di 0
E_M = -0.2

# 2. Calcoliamo r_max (il punto in cui V_eff = E_M)
# -G * M * m / r_max = E_M  ==>  r_max = -G * M * m / E_M
r_max = - (G * M * m) / E_M

# Stampa a video del valore di r_max calcolato
print(f"Il valore di r_max è: {r_max:.4f}")

# Creazione del grafico
plt.plot(r, V_r, label='Gravitational Potential Energy ($V_r$)')
plt.plot(r, E_M * np.ones_like(r), label='Total Energy ($E_M$)', linestyle='--')

# 3. Rappresentiamo l'intersezione sul grafico
# Disegna il punto di intersezione
plt.plot(r_max, E_M, 'ro', label=f'Intersezione')
# Disegna una linea verticale tratteggiata in corrispondenza di r_max
plt.axvline(r_max, color='red', linestyle=':', label='$r_{max}$')

plt.xlabel('Distance (r)')
plt.ylabel('Energy')
plt.title(r'Energy vs Distance $E_M < 0$')

# Ho modificato leggermente i limiti dell'asse y per inquadrare meglio E_M negativo
plt.ylim(-2, 0.2) 
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.legend()
plt.grid()

# Mantengo il percorso originale che avevi scelto per il salvataggio
plt.savefig("../../Appunti/img/energy_vs_distance_negativa.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Range di distanza modificato per inquadrare bene la buca e allontanarsi dall'asintoto in r=0
r = np.linspace(0.2, 30, 1000)

# 2. Unità "naturali" per rendere le forze comparabili sul grafico
G = 1
M = 10
m = 1

# Energia potenziale gravitazionale
V_r = - G * M * m / r

# 3. Momento angolare scelto appositamente per creare un minimo visibile
P = 7

# Energia potenziale centrifuga
V_c = P**2 / (2 * m * r**2)

# Energia potenziale efficace
V_eff = V_c + V_r

# --- NUOVO: Calcolo di r_c (minimo del potenziale) ---
# Derivata di V_eff rispetto a r posta a 0 porta a: r_c = P^2 / (G * M * m^2)
r_c = P**2 / (G * M * m**2)

# Calcoliamo anche il valore dell'energia (y) in quel punto per poterlo plottare
V_min = (P**2 / (2 * m * r_c**2)) - (G * M * m / r_c)
print(f"Il valore di r_c (minimo) è: {r_c:.2f}")
# -----------------------------------------------------

# Creazione del grafico
plt.figure(figsize=(8, 6)) # Opzionale: migliora le proporzioni

# Plot della curva principale (V_eff) in evidenza
plt.plot(r, V_eff, label='Effective Potential ($V_{eff}$)', color='blue', linewidth=2)

# Plot delle componenti (V_r e V_c) in secondo piano per mostrare come si forma V_eff
plt.plot(r, V_r, label='Gravitational Potential ($V_r$)', color='green', linestyle=':', alpha=0.7)
plt.plot(r, V_c, label='Centrifugal Potential ($V_c$)', color='orange', linestyle=':', alpha=0.7)

# --- NUOVO: Plottaggio del minimo r_c ---
# Punto rosso sul minimo
plt.plot(r_c, V_min, 'ro', label=f'Minimo)')
# Linea verticale tratteggiata per r_c
plt.axvline(r_c, color='red', linestyle=':', label='$r_c$')
# ----------------------------------------

plt.xlabel('Distance (r)')
plt.ylabel('Energy')
# Ho rimosso E_M dal titolo visto che in questo snippet specifico non lo stiamo plottando
plt.title('Energy vs Distance')

# 5. Limiti y adattati per inquadrare perfettamente il punto di minimo (che si trova in -1)
plt.ylim(-1.5, 2)

plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.legend()
plt.grid(True, alpha=0.4)

plt.savefig("../../Appunti/img/energy_vs_distance_pnonnullo.png")
plt.show()
```
