```python
import numpy as np
import matplotlib.pyplot as plt

a = 5.0
e = 0.7
c = a * e
p_param = a * (1 - e**2)

theta = np.linspace(0, 2 * np.pi, 1000)
r = p_param / (1 + e * np.cos(theta))
x = r * np.cos(theta)
y = r * np.sin(theta)

F2 = (0, 0)
F1 = (-2 * c, 0)

# Punto P generico
theta_p = np.pi / 2.5
r_p_gen = p_param / (1 + e * np.cos(theta_p))
P = (r_p_gen * np.cos(theta_p), r_p_gen * np.sin(theta_p))

# Calcolo Afelio e Perielio
# Perielio (distanza minima): theta = 0
r_per = p_param / (1 + e * 1)
Perielio = (r_per, 0) 

# Afelio (distanza massima): theta = pi, cos(pi) = -1
r_aph = p_param / (1 - e)
Afelio = (-r_aph, 0) 

plt.figure(figsize=(10, 7))

# Plot ellisse e linee del punto generico P
plt.plot(x, y, 'b-', linewidth=1.5, alpha=0.6)
plt.plot([F2[0], P[0]], [F2[1], P[1]], 'k--', linewidth=0.8, alpha=0.5)

# Evidenziazione asse per r_p e r_a
# Linea spessa arancione da F2 al Perielio (r_p)
plt.plot([0, Perielio[0]], [0, 0], color='orange', linewidth=4, alpha=0.8)
# Linea spessa viola da F2 all'Afelio (r_a)
plt.plot([0, Afelio[0]], [0, 0], color='purple', linewidth=4, alpha=0.8)

# Plot punti principali
plt.plot(F2[0], F2[1], 'ro', markersize=7, zorder=5) # F2 in primo piano
plt.plot(F1[0], F1[1], 'go', markersize=5, alpha=0.5) # F1 in secondo piano
plt.plot(P[0], P[1], 'ko', markersize=5)
plt.plot(Perielio[0], Perielio[1], 'o', color='orange', markersize=6)
plt.plot(Afelio[0], Afelio[1], 'o', color='purple', markersize=6)

# Label punti principali
plt.text(F2[0] - 0.2, F2[1] + 0.4, '$F_2$ (Fuoco)', fontsize=12, color='red', fontweight='bold', ha='right')
plt.text(F1[0], F1[1] + 0.4, '$F_1$', fontsize=10, color='green', ha='center')
plt.text(P[0] + 0.2, P[1] + 0.2, r'$P(r,\theta)$', fontsize=12)

# Staffe e Testo per le distanze r_p e r_a
bracket_y = -0.8

# Staffa r_p (Arancione)
plt.plot([0, 0], [bracket_y - 0.1, bracket_y + 0.1], color='orange', linewidth=1.5)
plt.plot([Perielio[0], Perielio[0]], [bracket_y - 0.1, bracket_y + 0.1], color='orange', linewidth=1.5)
plt.plot([0, Perielio[0]], [bracket_y, bracket_y], color='orange', linewidth=1.5)
plt.text(Perielio[0] / 2, bracket_y - 0.4, r'$r_p$ (Perielio)', ha='center', color='darkorange', fontsize=12, fontweight='bold')

# Staffa r_a (Viola)
plt.plot([Afelio[0], Afelio[0]], [bracket_y - 0.1, bracket_y + 0.1], color='purple', linewidth=1.5)
plt.plot([Afelio[0], 0], [bracket_y, bracket_y], color='purple', linewidth=1.5)
plt.text(Afelio[0] / 2, bracket_y - 0.4, r'$r_a$ (Afelio)', ha='center', color='purple', fontsize=12, fontweight='bold')

# Asse x di base per dare contesto
plt.axhline(0, color='black', linewidth=0.5, linestyle='-', alpha=0.3, zorder=0)

plt.gca().set_aspect('equal', adjustable='box')
plt.axis('off')
plt.tight_layout()
plt.savefig("../../Appunti/img/ellissePerielioAfelio.png", dpi = 300)
plt.show()
```
